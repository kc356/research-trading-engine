```python
import requests

r = requests.Session()

GAMMA = "https://gamma-api.polymarket.com"

slug = "btc-updown-5m-1785891000"

events = r.get(f"{GAMMA}/events", params={"slug": slug}).json()
```

```python
market = events[0].get("markets")[0]
market["outcomes"]
```
